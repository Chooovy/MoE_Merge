```python
import torch
import torch.nn as nn

def linear_cka(W1: torch.Tensor, W2: torch.Tensor) -> float:
    # 中心化
    def center_kernel(K):
        n = K.size(0)
        one_n = torch.ones(n, n) / n
        return K - one_n @ K - K @ one_n + one_n @ K @ one_n

    # 计算Gram矩阵
    K = W1 @ W1.T
    L = W2 @ W2.T

    # 中心化Gram矩阵
    Kc = center_kernel(K)
    Lc = center_kernel(L)

    # 计算CKA
    numerator = torch.trace(Kc @ Lc)
    denominator = torch.sqrt(torch.trace(Kc @ Kc) * torch.trace(Lc @ Lc))
    cka_score = numerator / denominator
    print(f"CKA between linear1 and linear2: {cka_score.item():.4f}")
    # return cka_score.item()
```
